```python
bilangan = int(input("masukan bilangan bulat: "))

if bilangan % 2 == 0:
  print("bilangan tersebut adalah genap")
else:
  print("bilangan tersebut adalah ganjil")
```

```python
nilai = int(input("masukan nilai: "))

if nilai >= 75:
   print("lulus")
else:
   print("Tidak Lulus")
```

```python
usia = int(input("Masukan usia: "))

if usia >= 18:
   print("Dewasa")
elif usia >= 13 and usia <= 17:
   print("Remaja")
else:
   print("Anak-anak")
```

```python
status = input("Masukan status keanggotaan: ")

if status == "gold" or status == "silver":
   print("Selamat! Anda mendapatkan diskon")
else:
   print("Kamu nga dapet diskon")
```

```python
jumlah = float(input("Masukan jumlah harga belanja: "))

if jumlah > 100:
   print("Total harga setelah diskon: ", - (jumlah * 10 / 100))
else:
   print("Total harga: ", jumlah)
```
